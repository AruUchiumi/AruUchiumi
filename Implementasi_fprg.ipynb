```python
print("Program Menghitung Selisih Tanggal")

tanggal1 = input("Masukkan tanggal pertama: ")
bulan1 = input("Masukkan bulan pertama: ")
tahun1 = input("Masukkan tahun pertama: ")
tanggal2 = input("Masukkan tanggal kedua: ")
bulan2 = input("Masukkan bulan kedua: ")
tahun2 = input("Masukkan tahun kedua: ")

total1 = int(tanggal1) + (int(bulan1)*30) + (int(tahun1)*365)
total2 = int(tanggal2) + (int(bulan2)*30) + (int(tahun2)*365)
selisih = abs(total1-total2)
tahun = selisih/365
sisa = selisih%365
bulan = sisa/30
hari = sisa%30

print("Selisih Tanggal adalah : ", int(tahun), "Tahun, ", int(bulan), "Bulan, ", int(hari), "Hari")
```

```python
print("Program Merotasikan Koordinat")

x = int(input("Masukkan nilai x yang ingin diubah : "))
y = int(input("Masukkan nilai y yang ingin diubah : "))
z = int(input("Masukkan nilai z yang ingin diubah : "))

temp = x
x = y
y = z
z = temp

print("Nilai x : ", x)
print("Nilai y : ", y)
print("Nilai z : ", z)
```
